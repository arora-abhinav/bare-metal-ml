```python
import pandas as pd
import numpy as np
from bare_metal_ml import linalg as custom_math
import math

#Row-column logic. Ex: Matrix[i][j] refers to row i and column j (indices)
#This is an implementation of Gaussian Discriminant Analysis (GDA), a generative learning
#algorithm that models p(x|y) as a multivariate Gaussian and uses Bayes rule to classify.
```

```python
df = pd.read_csv("wdbc.csv")

#Column 0 is the patient ID, column 1 is the diagnosis label (M/B), and columns 2 onwards
#are the 30 numerical features, which is why x starts at column 2.
x = df.iloc[:, 2:]
y = df.iloc[:, 1:2]

rows, cols = x.shape

#Shuffling before splitting avoids the case where the CSV's natural ordering puts examples
#of the same class together, which would make the train/test split non-representative.
shuffle_idx = np.random.permutation(rows)
train_size = int(0.8 * rows)

train_idx = shuffle_idx[:train_size]
test_idx = shuffle_idx[train_size:]

x_train = x.iloc[train_idx].values
x_test = x.iloc[test_idx].values
y_train = y.iloc[train_idx, 0].values
y_test = y.iloc[test_idx, 0].values

print(f"Training samples: {len(x_train)}, Test samples: {len(x_test)}")


dimension = x_train.shape
dimension_test = x_test.shape

print(x_train)
print(dimension)
print(y_test.shape)
```

```python
#Phi is p(y=1), the prior probability that a tumor is malignant. It is one of the
#parameters GDA learns from data, alongside mu_0, mu_1, and the shared covariance matrix.
def compute_phi_malignant():
    count = 0
    for i in range(train_size):
        if y_train[i] == 'M':
            count += 1

    return count/train_size

phi = compute_phi_malignant()
print(phi)
```

```python
#mu_zero is the mean feature vector for all benign (y=0) training examples. The denominator
#(1 - phi) * train_size is simply the total count of benign examples in the training set.
def mu_zero():
    init_arr = np.zeros(len(x_train[0]))
    for i in range(train_size):
        if y_train[i] != "M":
            for j in range(len(init_arr)):
                init_arr[j] += x_train[i][j]
    
    for k in range(len(init_arr)):
        init_arr[k] /= ((1-phi) * train_size)

    return init_arr.tolist()

mu_zero_arr = mu_zero()
```

```python
#mu_one follows the same logic as mu_zero but filters for malignant examples instead.
def mu_one():
    init_arr = np.zeros(len(x_train[0]))
    for i in range(train_size):
        if y_train[i] == "M":
            for j in range(len(init_arr)):
                init_arr[j] += x_train[i][j]
    
    for k in range(len(init_arr)):
        init_arr[k] /= (phi* train_size)

    return init_arr.tolist()

mu_one_arr = mu_one()

#GDA assumes both classes share a single covariance matrix, unlike QDA which fits a
#separate one per class. For each training example we compute the outer product
#(x - mu_y)(x - mu_y)^T using the mean of whichever class that example belongs to,
#and then average these outer products across all training examples.
def calculate_covariance_matrix(dimension):
    res = [[0] * dimension for _ in range(dimension)]

    for i in range(train_size):
        v2 = []
        if y_train[i] == 'M':
            v2 = mu_one_arr
        else:
            v2 = mu_zero_arr
        vector = custom_math.calculate_vector(x_train[i], v2)
        mat = custom_math.matrix_product_from_vector_and_transpose(dimension, vector)
        for j in range(dimension):
            for k in range(dimension):
                res[j][k] += (mat[j][k])/train_size
    
    return res
```

```python
#This computes p(x|y) using the multivariate Gaussian PDF. The covariance matrix is
#regularized before inversion to prevent it from being singular. LU decomposition is used
#to compute both the inverse and the determinant from the same factorization rather than
#doing them as two separate operations.
def multivariate_gaussian(mean, covariance_matrix, feature_vector, dimension):
    resultant_vector = custom_math.calculate_vector(feature_vector, mean)
    cov_copy = [row[:] for row in covariance_matrix]  # make a copy!
    cov_copy = custom_math.regularize(cov_copy, dimension)
    lower_triangular, upper_triangular = custom_math.LU_decomposition(cov_copy, dimension)
    inverse = custom_math.matrix_inverse(lower_triangular, upper_triangular, dimension)
    vector_prod = custom_math.matrix_product_with_matrix_and_vector(inverse, resultant_vector, dimension, dimension)
    scalar_prod = custom_math.scalar_product_from_transpose_and_vector(resultant_vector, vector_prod, dimension)
    determinant = custom_math.calculate_determinant(upper_triangular, dimension)
    exponential_term = math.exp(-0.5 * scalar_prod)
    denominator = 1/(((2 * math.pi)**(dimension/2)) * (abs(determinant) ** 0.5))
    res = denominator * exponential_term
    return res
```

```python
covar_mat = calculate_covariance_matrix(dimension[1])

#GDA classifies by comparing p(x|y=1)*phi against p(x|y=0)*(1-phi). We don't need the
#full posterior p(y|x) because p(x) is the same for both classes and cancels out.
def test_accuracy(dimension, x_test_len):
    correct = 0
    for i in range(x_test_len):
        pred = 0
        prediction_zero = multivariate_gaussian(mu_zero_arr, covar_mat, x_test[i].tolist(), dimension) * (1-phi)
        prediction_one = multivariate_gaussian(mu_one_arr, covar_mat, x_test[i].tolist(), dimension) * phi
        if(prediction_one > prediction_zero):
            pred = 1

        if(pred == 1 and y_test[i] == 'M'):
            correct += 1
        elif(pred == 0 and y_test[i] != 'M'):
            correct += 1
        
        res = 'M'
        if pred == 0:
            res = 'B'
            
        print(res)
        print(y_test[i])
        print("")
    
    return (correct/len(x_test)) * 100


print(test_accuracy(dimension[1], len(x_test)))
```
